# Import Dataset

```python
original_df = pd.read_csv('orders_messy.csv')

df = original_df.copy()
```

# Dataset

```python
df.head()
```

# Information About Dataset

```python
df.info()
# print(df.columns)
```

```python
df.describe()
```

# Missing Values

```python
df[['city','order_time','day_type','cuisine','meal_type','restaurant_type','discount_applied',
   'is_repeat_order','mood','hunger_level','company','rainy_weather']] = df[['city','order_time','day_type','cuisine',
                                                                                             'meal_type','restaurant_type','discount_applied',
                                                                                            'is_repeat_order','mood','hunger_level','company','rainy_weather']].fillna('Unknown')
```

# Not Important Columns

```python
df = df.drop(columns=['mood','hunger_level'])
```

```python
df['order_id'] = df['order_id'].abs()
```

# Replace the column category


```python
df['company'].value_counts()
```

```python
company_mapping = {
    'Partner': 'Zomato',
    'Friends': 'Swiggy',
    'Alone': 'Uber Eats',
    'Family': 'Domino\'s',
    'Unknown': 'Others'
}

df['company'] = df['company'].map(company_mapping)
```

```python
df['company'].value_counts()
```

#  Outlier Detection

```python
df.boxplot(column='Wind_Speed')
```

```python
df.sort_values("Wind_Speed", ascending=False).head(20)
```

```python
df['Wind_Speed'].value_counts()
```

```python
median_value = df['Wind_Speed'].median()

df['Wind_Speed'] = df['Wind_Speed'].replace(999999.0, median_value)
```

```python
df['Wind_Speed'].value_counts()
```

# Clean Dataset for Analysis

```python
df.to_csv('clean.csv',index=False)
```
